```python
import re
import pandas as pd
import gradio as gr
import plotly.express as px

from collections import Counter
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline


# ==========================================
# 1. MACHINE LEARNING MODEL
# ==========================================

TRAINING_DATA = {
    "Syntax Error": [
        "missing semicolon",
        "expected semicolon",
        "missing closing parenthesis",
        "expected closing brace",
        "unexpected token",
        "invalid syntax",
        "missing semicolon after declaration",
        "expected expression",
        "missing closing bracket",
        "expected parenthesis after if"
    ],
    "Semantic Error": [
        "undeclared variable",
        "variable is not declared",
        "duplicate declaration",
        "variable already declared",
        "type mismatch",
        "incompatible data types",
        "undefined identifier",
        "invalid variable usage",
        "using undeclared identifier",
        "redeclaration of variable"
    ],
    "Lexical Error": [
        "unknown symbol",
        "invalid character",
        "unrecognized token",
        "illegal character",
        "unknown lexical token",
        "invalid symbol in code"
    ]
}

training_texts = []
training_labels = []

for label, messages in TRAINING_DATA.items():
    for message in messages:
        training_texts.append(message)
        training_labels.append(label)

ml_model = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 2))),
    ("classifier", LogisticRegression(max_iter=1000))
])

ml_model.fit(training_texts, training_labels)


# ==========================================
# 2. LEXICAL ANALYSIS
# ==========================================

KEYWORDS = {
    "int", "float", "double", "char", "string",
    "bool", "void", "if", "else", "while", "for",
    "return", "main", "cout", "cin", "endl",
    "using", "namespace", "std", "const"
}

TOKEN_PATTERN = re.compile(
    r'//.*|'
    r'"(?:\\.|[^"\\])*"|'
    r"'(?:\\.|[^'\\])*'|"
    r'\d+(?:\.\d+)?|'
    r'[A-Za-z_]\w*|'
    r'<<|>>|==|!=|<=|>=|\+\+|--|&&|\|\||'
    r'[+\-*/%=<>!&|^~]|'
    r'[;,:(){}\[\].]'
)


def lexical_analysis(code):
    tokens = []
    errors = []

    for line_no, line in enumerate(code.splitlines(), 1):
        if line.strip().startswith("#"):
            continue

        pos = 0

        while pos < len(line):
            if line[pos].isspace():
                pos += 1
                continue

            match = TOKEN_PATTERN.match(line, pos)

            if not match:
                errors.append({
                    "Line": line_no,
                    "Stage": "Lexical",
                    "Error": f"Unknown symbol '{line[pos]}'",
                    "Suggestion": "Check the character."
                })
                pos += 1
                continue

            value = match.group(0)

            if value.startswith("//"):
                break

            if value.startswith('"') or value.startswith("'"):
                category = "String/Character"
            elif value[0].isdigit():
                category = "Number"
            elif re.fullmatch(r'[A-Za-z_]\w*', value):
                category = (
                    "Keyword" if value in KEYWORDS
                    else "Identifier"
                )
            elif value in ";,:(){}[]":
                category = "Delimiter"
            else:
                category = "Operator"

            tokens.append({
                "Line": line_no,
                "Token": value,
                "Category": category
            })

            pos = match.end()

    return tokens, errors


# ==========================================
# 3. SYNTAX ANALYSIS
# ==========================================

def syntax_analysis(code):
    errors = []
    stack = []
    pairs = {")": "(", "}": "{", "]": "["}

    for line_no, line in enumerate(code.splitlines(), 1):
        if line.strip().startswith("#"):
            continue

        clean = re.sub(
            r'"(?:\\.|[^"\\])*"|\'(?:\\.|[^\'\\])*\'',
            '""', line
        ).split("//", 1)[0]

        for char in clean:
            if char in "({[":
                stack.append((char, line_no))
            elif char in ")}]":
                if stack and stack[-1][0] == pairs[char]:
                    stack.pop()
                else:
                    errors.append({
                        "Line": line_no,
                        "Stage": "Parsing",
                        "Error": f"Unexpected '{char}'",
                        "Suggestion": "Check matching brackets."
                    })

        statement = clean.strip()

        if not statement:
            continue

        needs_semicolon = (
            re.match(
                r'^(int|float|double|char|string|bool)\b',
                statement
            )
            or statement.startswith("cout ")
            or statement.startswith("cin ")
            or statement.startswith("return ")
        )

        if needs_semicolon and not statement.endswith(";"):
            errors.append({
                "Line": line_no,
                "Stage": "Parsing",
                "Error": "Missing semicolon",
                "Suggestion": "Add ';' at the end."
            })

        if re.match(r'^int\s*=', statement):
            errors.append({
                "Line": line_no,
                "Stage": "Parsing",
                "Error": "Missing variable name",
                "Suggestion": "Add an identifier after int."
            })

    for opening, line_no in stack:
        closing = {"(": ")", "{": "}", "[": "]"}[opening]

        errors.append({
            "Line": line_no,
            "Stage": "Parsing",
            "Error": f"Missing closing '{closing}'",
            "Suggestion": f"Add '{closing}'."
        })

    return errors


# ==========================================
# 4. SEMANTIC ANALYSIS
# ==========================================

def semantic_analysis(code):
    errors = []
    symbols = {}

    declaration = re.compile(
        r'\b(int|float|double|char|string|bool)\s+'
        r'([A-Za-z_]\w*)'
    )

    # Build the symbol table.
    for line_no, line in enumerate(code.splitlines(), 1):
        if line.strip().startswith("#"):
            continue

        line = line.split("//", 1)[0]

        for match in declaration.finditer(line):
            dtype, name = match.groups()

            if name in symbols:
                errors.append({
                    "Line": line_no,
                    "Stage": "Semantic",
                    "Error": f"Duplicate declaration: {name}",
                    "Suggestion": "Rename or remove the duplicate."
                })
            else:
                symbols[name] = {
                    "Data Type": dtype,
                    "Declaration Line": line_no
                }

    # Check basic identifier usage.
    for line_no, line in enumerate(code.splitlines(), 1):
        if line.strip().startswith("#"):
            continue

        line = line.split("//", 1)[0]

        line = re.sub(
            r'"(?:\\.|[^"\\])*"|\'(?:\\.|[^\'\\])*\'',
            ' ', line
        )

        # Remove declaration names from the usage check.
        line = declaration.sub(
            lambda m: m.group(1) + " ",
            line
        )

        for name in re.findall(r'\b[A-Za-z_]\w*\b', line):
            if name in KEYWORDS or name in symbols:
                continue

            errors.append({
                "Line": line_no,
                "Stage": "Semantic",
                "Error": f"Undeclared identifier: {name}",
                "Suggestion": f"Declare '{name}' before use."
            })

            # Prevent duplicate reports in this basic prototype.
            symbols[name] = {
                "Data Type": "Unknown",
                "Declaration Line": "Not declared"
            }

    symbol_rows = [
        {"Variable": name, **details}
        for name, details in symbols.items()
    ]

    return errors, symbol_rows


# ==========================================
# 5. CORRECTION ENGINE
# ==========================================

def apply_corrections(code, errors):
    lines = code.splitlines()
    changed = []

    for error in errors:
        line_no = error["Line"]

        if error["Error"] != "Missing semicolon":
            continue

        if 1 <= line_no <= len(lines):
            original = lines[line_no - 1]

            if original.strip() and not original.rstrip().endswith(
                (";", "{", "}", ":")
            ):
                lines[line_no - 1] = original.rstrip() + ";"
                changed.append(line_no)

    return "\n".join(lines), changed


# ==========================================
# 6. COMPLETE ANALYSIS FUNCTION
# ==========================================

def analyze_code(code):
    if not code or not code.strip():
        return (
            "Please enter some C++ code.",
            pd.DataFrame(),
            pd.DataFrame(),
            pd.DataFrame(),
            "",
            None
        )

    tokens, lexical_errors = lexical_analysis(code)
    syntax_errors = syntax_analysis(code)
    semantic_errors, symbols = semantic_analysis(code)

    all_errors = lexical_errors + syntax_errors + semantic_errors

    for error in all_errors:
        prediction = ml_model.predict([error["Error"]])[0]
        scores = ml_model.predict_proba([error["Error"]])[0]

        error["ML Prediction"] = prediction
        error["Model Score (%)"] = round(float(max(scores) * 100), 2)

    corrected, changed = apply_corrections(code, all_errors)

    summary = (
        f"### Analysis Summary\n"
        f"- Tokens found: **{len(tokens)}**\n"
        f"- Lexical errors: **{len(lexical_errors)}**\n"
        f"- Syntax errors: **{len(syntax_errors)}**\n"
        f"- Semantic errors: **{len(semantic_errors)}**\n"
        f"- Total potential issues: **{len(all_errors)}**\n\n"
    )

    if all_errors:
        summary += "Review the error table and suggestions below."
    else:
        summary += "No issues were detected by the implemented checks."

    error_df = pd.DataFrame(all_errors)
    token_df = pd.DataFrame(tokens)
    symbol_df = pd.DataFrame(symbols)

    download_file = None

    if changed:
        download_file = "corrected_program.cpp"
        with open(download_file, "w", encoding="utf-8") as file:
            file.write(corrected)

    return (
        summary,
        error_df,
        token_df,
        symbol_df,
        corrected,
        download_file
    )


print("Compiler and ML components are ready.")
```

```python
CSS = """
body {
    background: #0b1220 !important;
}

.gradio-container {
    max-width: 1400px !important;
    background: #0b1220 !important;
    color: #e7eefb !important;
}

#hero {
    background: linear-gradient(120deg, #193457, #111c30);
    border: 1px solid #2d4566;
    padding: 24px;
    border-radius: 18px;
    margin-bottom: 16px;
}

#hero h1 {
    color: white;
    font-size: 34px;
}

#hero p {
    color: #bfd0e8;
}

button.primary {
    background: #3478f6 !important;
    color: white !important;
}
"""

DEFAULT_CODE = """#include <iostream>
using namespace std;

int main() {
    int a = 10
    int b = 20;
    cout << a + b;
    return 0;
}"""

with gr.Blocks(
    theme=gr.themes.Soft(
        primary_hue="blue",
        neutral_hue="slate"
    ),
    css=CSS,
    title="CodeGuard AI"
) as demo:

    gr.HTML("""
    <div id="hero">
      <h1>🧠 CodeGuard AI</h1>
      <p>AI-Enhanced Compiler for Detecting and Correcting Programming Errors</p>
      <p>Lexical Analysis · Parsing · Semantic Analysis · ML Prediction</p>
    </div>
    """)

    with gr.Row():
        with gr.Column(scale=3):
            code_input = gr.Code(
                label="💻 C++ Source Code",
                language="cpp",
                value=DEFAULT_CODE,
                lines=18
            )

            with gr.Row():
                analyze_button = gr.Button(
                    "🔍 Analyze Code",
                    variant="primary"
                )
                example_button = gr.Button("↻ Load Example")

        with gr.Column(scale=2):
            summary_output = gr.Markdown(
                "### Ready to analyze\nEnter code and click Analyze Code."
            )

            corrected_output = gr.Code(
                label="🛠️ Corrected Code Preview",
                language="cpp",
                lines=12
            )

            download_output = gr.File(
                label="⬇️ Download Corrected C++ File"
            )

    with gr.Tabs():
        with gr.Tab("🚨 Error Report"):
            errors_output = gr.Dataframe(
                label="Detected Errors and ML Predictions",
                interactive=False
            )

        with gr.Tab("🔬 Token Explorer"):
            tokens_output = gr.Dataframe(
                label="Tokens",
                interactive=False
            )

        with gr.Tab("📚 Symbol Table"):
            symbols_output = gr.Dataframe(
                label="Declared Variables",
                interactive=False
            )

        with gr.Tab("ℹ️ About Project"):
            gr.Markdown("""
            ## Project Information

            **Title:** AI-Enhanced Compiler for Detecting and Correcting Programming Errors

            **Compiler Design**
            - Lexical analysis
            - Parsing and syntax checks
            - Symbol table and semantic checks

            **Machine Learning**
            - TF-IDF feature extraction
            - Logistic Regression error classification

            **Team Registration Numbers**
            - 24BCB7054
            - 24BCB7058

            **Limitations:** This prototype checks a subset of C++ syntax.
            ML scores are not validated real-world accuracy.
            """)

    outputs = [
        summary_output,
        errors_output,
        tokens_output,
        symbols_output,
        corrected_output,
        download_output
    ]

    analyze_button.click(
        fn=analyze_code,
        inputs=code_input,
        outputs=outputs
    )

    example_button.click(
        fn=lambda: DEFAULT_CODE,
        inputs=None,
        outputs=code_input
    )

print("Interface created.")
```
